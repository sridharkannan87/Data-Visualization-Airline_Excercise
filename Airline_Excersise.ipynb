## Creayed a function that returns a rank (Best, Good, Fair) based on customer online feedback

```python
def calc(OS):
    if (OS>=4):
        return 'Best'
    elif (OS>=2):
        return 'Good'
    else: return 'Fair'
  
        
```

```python
airl_data_df['Ranking']=airl_data_df['On-board service'].apply(calc)
```
